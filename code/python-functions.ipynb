## Einfache Funktion

Beginnen wir mit einer einfachen Funktion, die das Quadrat einer Zahl berechnet:

```python
# Funktion zur Berechnung des Quadrats einer Zahl
def square(num):
    """Berechnen und zurückgeben des Quadrats einer Zahl."""
    return num * num

# Verwendung der Funktion zur Berechnung des Quadrats von 4
result = square(4)
print(f"Quadrat von 4: {result}")
```

Erläuterung:

- `def` beginnt die Funktionsdefinition, gefolgt vom Funktionsnamen `square` und einem Parameter `num`.
- Die Docstring `"""` beschreibt kurz den Zweck der Funktion.
- Die Funktion berechnet das Quadrat mit der Formel `num * num`.
- Die `return`-Anweisung gibt das Ergebnis an die Stelle zurück, an der die Funktion aufgerufen wurde.
- Die `print`-Funktion verwendet einen f-String, eine formatierte Zeichenkette, die mit `f` vorangestellt ist und das Einfügen von Variablen innerhalb von `{}` geschweiften Klammern ermöglicht. Sie ist kürzer und leichter lesbar als die traditionelle Zeichenkettenverkettung.

Vorteile von f-Strings:

- Sie bieten eine saubere, intuitive Möglichkeit, Variablen in Zeichenketten einzufügen.
- Unterstützung für Formatierungen ist eingebaut (z.B. Dezimalstellen, Auffüllungen).
- Sie vereinfachen die Zeichenketteninterpolation und verbessern die Lesbarkeit.

## Funktion mit mehreren Parametern

Funktionen können mehr als einen Parameter akzeptieren. Hier ist eine Funktion, die die Fläche eines Rechtecks berechnet:

```python
# Funktion zur Berechnung der Fläche eines Rechtecks
def calculate_rectangle_area(length, width):
    """Berechnen der Fläche eines Rechtecks anhand seiner Länge und Breite."""
    return length * width

# Verwendung der Funktion mit Beispielwerten
length = 10
width = 5

area = calculate_rectangle_area(length, width)

print(f"Rechteck mit Länge {length} und Breite {width} hat eine Fläche von {area}")
```

Erläuterung:

- `length` und `width` sind Parameter, die die Funktion zur Berechnung der Fläche des Rechtecks verwendet.
- `calculate_rectangle_area` wird mit bestimmten Werten aufgerufen.
- Der f-String innerhalb der `print`-Anweisung liefert eine klare Ausgabe, die sowohl die Eingabewerte als auch das berechnete Ergebnis zeigt.

## Standardparameter

Standardwerte für Parameter werden verwendet, wenn keine Argumente bereitgestellt werden.

```python
# Funktion mit einem Standardparameterwert
def greet(name="there"):
    """Begrüßen einer Person mit ihrem Namen oder mit 'Hallo zusammen' als Standard."""
    print(f"Hallo, {name}!")

# Aufruf der Funktion mit einem bestimmten Namen
greet("Alice")

# Aufruf der Funktion ohne Argument (verwendet den Standardwert)
greet()
```

Erläuterung:

- Der Parameter `name="there"` liefert einen Standardwert.
- Beim Aufruf von `greet("Alice")` wird "Hallo, Alice!" ausgegeben.
- Der Aufruf von `greet()` ohne Argumente verwendet den Standardwert "there".

## Lambda-Funktionen

Lambda-Funktionen sind kleine, anonyme Funktionen, die nützlich für kurze Operationen oder das Übergeben als Argumente sind. Hier ist ein einfaches Beispiel:

```python
# Lambda-Funktion zum Addieren zweier Zahlen
add = lambda x, y: x + y

# Verwendung der Lambda-Funktion
result = add(3, 5)
print(f"Summe von 3 und 5: {result}")
```

Erläuterung:

- `lambda` definiert eine schnelle Funktion in einer einzigen Zeile.
- Der Ausdruck `x + y` wird berechnet und zurückgegeben.
- Lambda-Funktionen sind nützlich, wenn Sie eine einfache Funktion vorübergehend benötigen.

## Lambda-Funktionen mit DataFrames

Lambda-Funktionen sind besonders nützlich, wenn Sie eine benutzerdefinierte Transformation auf jede Zeile oder Spalte eines DataFrames anwenden möchten.

Zum Beispiel, wenn Sie Zahlen basierend auf bestimmten Bedingungen klassifizieren möchten, kann eine Lambda-Funktion helfen.

Beispielcode:

```python
# Erstellen eines DataFrame und Verwendung einer Lambda-Funktion für eine benutzerdefinierte Klassifizierung
import pandas as pd

# Beispiel-DataFrame
data = {"Numbers": [1, 2, 3, 4, 5, 15, 20, 25]}
df = pd.DataFrame(data)

# Definieren einer benutzerdefinierten Funktion zur Klassifizierung von Zahlen
def classify_number(num):
    if num < 10:
        return "Einstellig"
    elif num < 20:
        return "Teenager"
    else:
        return "Zwanzig oder mehr"

# Anwenden der benutzerdefinierten Funktion mit einer Lambda-Funktion zur Erstellung einer neuen Spalte
df["Klassifizierung"] = df["Numbers"].apply(lambda x: classify_number(x))

# Anzeigen des DataFrame
df
```
